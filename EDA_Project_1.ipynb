```python
#To find all rows involved in a duplicate
import pandas as pd
df = pd.read_csv('/adult.csv')
print(df)
duplicates=df[df.duplicated(keep=False)]
print(duplicates)
```

```python
df.duplicated().sum()
```

```python
#To remove duplicate rows
df=df.drop_duplicates()
print(df)
```

```python
#For Filling NUll values[ffill-forward filling, bfill-backward filling, interpolate(exact values which suits there)]
df["Workclass"] = df["Workclass"].ffill()
print(df)
```

```python
#Detecting Outliers (Q1,Q2,Q3,IQR)
Q1 = df['Age'].quantile(0.25)
Q3 = df['Age'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1-1.5*IQR  #Formula For LowerBound
upper_bound = Q3+1.5*IQR  #Formula For UpperBound
print(lower_bound)
print(upper_bound)
```

```python
#Identifying Outliers
iqr_outliers = df[
    (df['Age'] < lower_bound) |
    (df['Age'] > upper_bound)
]
print("IQR Outliers:", len(iqr_outliers))
```

```python
#Z-score Method
from scipy.stats import zscore
df['Age_zscore'] = zscore(df['Age'])
z_outliers = df[
    df['Age_zscore'].abs() > 3
]
print("Z-Score Outliers:", len(z_outliers))
```

```python
#Z-Score Trimming
df_trimmed = df[df['Age_zscore'].abs() <= 3]
print(df_trimmed)
```

```python
# Winsorization
from scipy.stats.mstats import winsorize
df['Age_Winsorized'] = winsorize(
    df['Age'],
    limits=[0,0.3]
)
print(df.head())
```

```python
#Detecting outliers for EducationNum
Q1 = df['EducationNum'].quantile(0.25)
Q2 = df['EducationNum'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5*IQR
upper = Q3 + 1.5*IQR
print(lower)
print(upper)
```

```python
#Identifying Outliers
iqr_outliers = df[
    (df['EducationNum'] < lower) |
    (df['EducationNum'] > upper)
]
print("IQR Outliers:", len(iqr_outliers))
```

```python
#Detecting outliers for Hours per Week
Q1 = df['Hours per Week'].quantile(0.25)
Q3 = df['Hours per Week'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5*IQR
upper_bound = Q1 + 1.5*IQR
print(lower_bound)
print(upper_bound)
```

```python
iqr_outliers = df[
    (df['Hours per Week'] < lower) |
    (df['Hours per Week'] > upper)
]
print("IQR Outliers:", len(iqr_outliers))
```

```python
#Visualization
#Boxplot - To remove Outliers from Age
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv("/adult.csv")
sns.boxplot(x=df["Age"])
plt.title("Age Box Plot")
plt.xlabel("Age")
plt.show()
```

```python
#Scatter Plot - Age vs Hours per week
plt.scatter(df["Age"], df["Hours per Week"])
plt.title("Age vs Hours per Week")
plt.xlabel("Age")
plt.ylabel("Hours per Week")
plt.show()
```

```python
#Bar plot
education = df["Education"].value_counts()
plt.bar(education.index, education.values)
plt.title("Education Distribution")
plt.xlabel("Education")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()
```

```python
#Heatmap
data = df[["Age", "EducationNum", "Capital Gain", "capital loss", "Hours per Week"]].values
plt.imshow(data, cmap="Greens")  # cmap is used for colours
plt.title("Adult Dataset Heatmap")
plt.xlabel("Columns")
plt.ylabel("Students")
plt.colorbar()
plt.show()
```
